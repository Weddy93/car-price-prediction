```python

# Inspect the cleaned dataset before model preparation
print("Dataset shape:", df_clean.shape)

print("\nColumn names:")
print(df_clean.columns.tolist())

print("\nData types:")
print(df_clean.dtypes)

print("\nMissing values:")
print(df_clean.isnull().sum())

print("\nTarget variable summary:")
print(df_clean["selling_price"].describe())

```

```python
#PRE PROCESSING THE PIPELINE

# Select the features used for prediction
feature_columns = [
    "brand",
    "car_age",
    "km_driven",
    "fuel",
    "seller_type",
    "transmission",
    "owner"
]

X = df_clean[feature_columns]
y = df_clean["selling_price"]

print("Features shape:", X.shape)
print("Target shape:", y.shape)

print("\nFeature preview:")
display(X.head())

print("\nTarget preview:")
display(y.head())
```

```python
#data splitting

from sklearn.model_selection import train_test_split

X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size=0.20,
    random_state=42
)

print("Training features:", X_train.shape)
print("Testing features:", X_test.shape)
print("Training targets:", y_train.shape)
print("Testing targets:", y_test.shape)
```

```python
#building the pipeline

from sklearn.compose import ColumnTransformer
from sklearn.preprocessing import OneHotEncoder
from sklearn.pipeline import Pipeline

categorical_features = [
    "brand",
    "fuel",
    "seller_type",
    "transmission",
    "owner"
]

numerical_features = [
    "car_age",
    "km_driven"
]

preprocessor = ColumnTransformer(
    transformers=[
        (
            "categorical",
            OneHotEncoder(handle_unknown="ignore"),
            categorical_features
        ),
        (
            "numerical",
            "passthrough",
            numerical_features
        )
    ]
)

print("Preprocessor created successfully.")
```

```python
#TRAINING THE BASELINE MODEL
#model pipeline

from sklearn.linear_model import LinearRegression
from sklearn.pipeline import Pipeline

linear_model = Pipeline(
    steps=[
        ("preprocessor", preprocessor),
        ("regressor", LinearRegression())
    ]
)

print("Linear Regression pipeline created successfully.")

# Train the baseline model
linear_model.fit(X_train, y_train)

print("Linear Regression training completed.")

```

```python

from sklearn.metrics import mean_absolute_error, mean_squared_error, r2_score
import numpy as np

# Predict prices for unseen test records
y_pred = linear_model.predict(X_test)

# Calculate evaluation metrics
mae = mean_absolute_error(y_test, y_pred)
rmse = np.sqrt(mean_squared_error(y_test, y_pred))
r2 = r2_score(y_test, y_pred)

print(f"MAE: ₹{mae:,.2f}")
print(f"RMSE: ₹{rmse:,.2f}")
print(f"R² Score: {r2:.4f}")
```

```python
from sklearn.tree import DecisionTreeRegressor
from sklearn.pipeline import Pipeline
from sklearn.metrics import mean_absolute_error, mean_squared_error, r2_score
import numpy as np

# Create the Decision Tree Regressor Pipeline
tree_model = Pipeline(
    steps=[
        ("preprocessor", preprocessor),
        ("regressor", DecisionTreeRegressor(random_state=42))
    ]
)

# Train the Decision Tree model
tree_model.fit(X_train, y_train)
print("Decision Tree training completed successfully.\n")

# Predict prices on the test set
tree_pred = tree_model.predict(X_test)

# Calculate evaluation metrics
tree_mae = mean_absolute_error(y_test, tree_pred)
tree_rmse = np.sqrt(mean_squared_error(y_test, tree_pred))
tree_r2 = r2_score(y_test, tree_pred)

print(f"Decision Tree MAE: ₹{tree_mae:,.2f}")
print(f"Decision Tree RMSE: ₹{tree_rmse:,.2f}")
print(f"Decision Tree R² Score: {tree_r2:.4f}")
```

```python
#comparing the two models

import pandas as pd

comparison = pd.DataFrame({
    "Model": ["Linear Regression", "Decision Tree"],
    "MAE": [mae, tree_mae],
    "RMSE": [rmse, tree_rmse],
    "R2 Score": [r2, tree_r2]
})

display(comparison.round(4))
```

```python
#RANDOM FOREST REGRESSOR


from sklearn.ensemble import RandomForestRegressor
from sklearn.pipeline import Pipeline

random_forest_model = Pipeline(
    steps=[
        ("preprocessor", preprocessor),
        (
            "regressor",
            RandomForestRegressor(
                n_estimators=200,
                max_depth=15,
                min_samples_leaf=2,
                random_state=42,
                n_jobs=-1
            )
        )
    ]
)

random_forest_model.fit(X_train, y_train)

print("Random Forest training completed successfully.")

#evaluating the model

rf_pred = random_forest_model.predict(X_test)

rf_mae = mean_absolute_error(y_test, rf_pred)
rf_rmse = np.sqrt(mean_squared_error(y_test, rf_pred))
rf_r2 = r2_score(y_test, rf_pred)

print(f"Random Forest MAE: ₹{rf_mae:,.2f}")
print(f"Random Forest RMSE: ₹{rf_rmse:,.2f}")
print(f"Random Forest R² Score: {rf_r2:.4f}")


#comparing the three models

rf_pred = random_forest_model.predict(X_test)

rf_mae = mean_absolute_error(y_test, rf_pred)
rf_rmse = np.sqrt(mean_squared_error(y_test, rf_pred))
rf_r2 = r2_score(y_test, rf_pred)

print(f"Random Forest MAE: ₹{rf_mae:,.2f}")
print(f"Random Forest RMSE: ₹{rf_rmse:,.2f}")
print(f"Random Forest R² Score: {rf_r2:.4f}")

```

```python

comparison = pd.DataFrame({
    "Model": [
        "Linear Regression",
        "Decision Tree",
        "Random Forest"
    ],
    "MAE": [mae, tree_mae, rf_mae],
    "RMSE": [rmse, tree_rmse, rf_rmse],
    "R2 Score": [r2, tree_r2, rf_r2]
})

display(
    comparison.sort_values("RMSE").reset_index(drop=True).round(4)
)
```

```python
#cross validation

from sklearn.model_selection import KFold, cross_validate
from sklearn.ensemble import RandomForestRegressor
from sklearn.compose import ColumnTransformer
from sklearn.preprocessing import OneHotEncoder
from sklearn.pipeline import Pipeline
import pandas as pd
import numpy as np

# Ensure dataset and variables are loaded and processed
try:
    _ = df_clean
except NameError:
    df_new = pd.read_csv("CAR DETAILS FROM CAR DEKHO.csv")
    df_clean = df_new.drop_duplicates().reset_index(drop=True)
    df_clean["brand"] = df_clean["name"].str.strip().str.split().str[0].str.title()
    df_clean["brand"] = df_clean["brand"].replace({"Land": "Land Rover", "Opelcorsa": "Opel", "Bmw": "BMW", "Mg": "MG"})
    df_clean["car_age"] = 2026 - df_clean["year"]

# Re-define features and target
feature_columns = ["brand", "car_age", "km_driven", "fuel", "seller_type", "transmission", "owner"]
X = df_clean[feature_columns]
y = df_clean["selling_price"]

# Split data (using the same random state 42 for consistency)
from sklearn.model_selection import train_test_split
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.20, random_state=42)

# Re-define preprocessor
categorical_features = ["brand", "fuel", "seller_type", "transmission", "owner"]
numerical_features = ["car_age", "km_driven"]

preprocessor = ColumnTransformer(
    transformers=[
        ("categorical", OneHotEncoder(handle_unknown="ignore"), categorical_features),
        ("numerical", "passthrough", numerical_features)
    ]
)

# Instantiate the random_forest_model pipeline
random_forest_model = Pipeline(
    steps=[
        ("preprocessor", preprocessor),
        (
            "regressor",
            RandomForestRegressor(
                n_estimators=200,
                max_depth=15,
                min_samples_leaf=2,
                random_state=42,
                n_jobs=-1
            )
        )
    ]
)

cv = KFold(
    n_splits=5,
    shuffle=True,
    random_state=42
)

cv_results = cross_validate(
    random_forest_model,
    X_train,
    y_train,
    cv=cv,
    scoring={
        "mae": "neg_mean_absolute_error",
        "rmse": "neg_root_mean_squared_error",
        "r2": "r2"
    },
    n_jobs=-1
)

cv_mae = -cv_results["test_mae"]
cv_rmse = -cv_results["test_rmse"]
cv_r2 = cv_results["test_r2"]

print(f"Mean CV MAE: ‹{cv_mae.mean():,.2f}")
print(f"Mean CV RMSE: ‹{cv_rmse.mean():,.2f}")
print(f"Mean CV R²: {cv_r2.mean():.4f}")

print(f"CV MAE standard deviation: ‹{cv_mae.std():,.2f}")
print(f"CV RMSE standard deviation: ‹{cv_rmse.std():,.2f}")
```

```python
#hyperparameter tuning

from sklearn.model_selection import RandomizedSearchCV
from sklearn.ensemble import RandomForestRegressor

param_distributions = {
    "regressor__n_estimators": [100, 200, 300, 400],
    "regressor__max_depth": [None, 10, 15, 20, 25],
    "regressor__min_samples_split": [2, 5, 10],
    "regressor__min_samples_leaf": [1, 2, 4, 6],
    "regressor__max_features": [0.7, 1.0]
}

random_search = RandomizedSearchCV(
    estimator=random_forest_model,
    param_distributions=param_distributions,
    n_iter=20,
    cv=cv,
    scoring="neg_mean_absolute_error",
    random_state=42,
    n_jobs=-1,
    verbose=1
)

random_search.fit(X_train, y_train)

print("Hyperparameter tuning completed.")
print("Best parameters:")
print(random_search.best_params_)
print(f"Best mean CV MAE: ₹{-random_search.best_score_:,.2f}")
```

```python
#evaluation of tuned model

best_model = random_search.best_estimator_

tuned_pred = best_model.predict(X_test)

tuned_mae = mean_absolute_error(y_test, tuned_pred)
tuned_rmse = np.sqrt(mean_squared_error(y_test, tuned_pred))
tuned_r2 = r2_score(y_test, tuned_pred)

print(f"Tuned Random Forest MAE: ₹{tuned_mae:,.2f}")
print(f"Tuned Random Forest RMSE: ₹{tuned_rmse:,.2f}")
print(f"Tuned Random Forest R² Score: {tuned_r2:.4f}")
```

```python

import re

def extract_car_model(name):
    parts = str(name).strip().split()

    if len(parts) < 2:
        return "Unknown"

    # Handle multi-word brands
    if parts[0].lower() == "mercedes-benz":
        remainder = parts[1:]
    elif len(parts) >= 2 and " ".join(parts[:2]).lower() in {
        "land rover"
    }:
        remainder = parts[2:]
    else:
        remainder = parts[1:]

    if not remainder:
        return "Unknown"

    # Handle common model names containing multiple words
    lower_parts = [p.lower() for p in remainder]

    known_patterns = [
        (["5", "series"], "5 Series"),
        (["3", "series"], "3 Series"),
        (["7", "series"], "7 Series"),
        (["wagon", "r"], "Wagon R"),
        (["grand", "i10"], "Grand i10"),
        (["grand", "vitara"], "Grand Vitara"),
        (["swift", "dzire"], "Swift Dzire"),
        (["alto", "800"], "Alto 800"),
        (["new", "c-class"], "C-Class"),
        (["new", "e-class"], "E-Class"),
    ]

    for pattern, model in known_patterns:
        if lower_parts[:len(pattern)] == pattern:
            return model

    # Default: first token, with basic cleanup
    return remainder[0].title()


df_model_v3 = df_clean.copy()

df_model_v3["car_model_v3"] = df_model_v3["name"].apply(
    extract_car_model
)

display(
    df_model_v3[
        ["name", "brand", "car_model_v3"]
    ].sample(20, random_state=42)
)
```
